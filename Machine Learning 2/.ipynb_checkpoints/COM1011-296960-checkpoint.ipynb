(7) Load the fashion-MNIST test dataset (`t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`), then obtain the image data denoted as `X_test` and the ground truth labels denoted as `y_test`. After that, pre-process `X_test` by scaling the image pixel values to values between 0 and 1. Source code is provided. [0 mark]

[hint: check https://github.com/zalandoresearch/fashion-mnist?tab=readme-ov-file#get-the-data for detailed information about fashion-MNIST dataset]


```python
# answer to (7). You may need to change path for `t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

def load_mnist(path, kind='train'):
    import os
    import gzip
    import numpy as np

    """Load fashion-MNIST data from `path`"""
    labels_path = os.path.join(path,
                               '%s-labels-idx1-ubyte.gz'
                               % kind)
    images_path = os.path.join(path,
                               '%s-images-idx3-ubyte.gz'
                               % kind)

    with gzip.open(labels_path, 'rb') as lbpath:
        labels = np.frombuffer(lbpath.read(), dtype=np.uint8,
                               offset=8)

    with gzip.open(images_path, 'rb') as imgpath:
        images = np.frombuffer(imgpath.read(), dtype=np.uint8,
                               offset=16).reshape(len(labels), 784)

    return images, labels


X_test, y_test = load_mnist('t10k files', kind='t10k')
X_train, y_train = load_mnist("t10k files", kind="train")

X_test = X_test/255

print(np.max(X_test))
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)

```

(8) Obtain all the test images and their (ground-truth) labels for '3','6','7', respectively, and then rearrange the samples and labels in a random order. [1 mark]

[hint: check https://numpy.org/doc/2.2/reference/generated/numpy.isin.html and https://numpy.org/doc/2.1/reference/random/generated/numpy.random.permutation.html]



```python
# answer to (8)
correct_values = [3,6,7]
true_values = np.isin(y_test, correct_values)
image_list = X_test[true_values]
label_list = y_test[true_values]

permutation = np.random.permutation(len(label_list))

random_test_label = label_list[permutation]
random_test_images = image_list[permutation]
        
        
```

(9) Use PCA with 2 components to visualise the test data samples in (8) with different colors representing different real (i.e., ground-truth) clusters in a 2d plot. In the plot legend, indicate which colours correspond to which digits. You should use three colors for the three labels: '3','6', '7'. [2 marks]

```python
#Answer to (9)
#perform PCA with 2 components and further visualize the data samples in a 2d plot
from sklearn.decomposition import PCA

pca_model = PCA(n_components = 2)
images_pca = pca_model.fit_transform(random_test_images)

plt.figure(figsize =(8,6)) 

for digit in [3,6,7]:
    mask = random_test_label == digit
    colour = np.random.rand(3,)
    plt.scatter(images_pca[mask, 0],images_pca[mask, 1], label=str(digit),color = colour , alpha=0.6)

plt.xlabel("Principle component 1")
plt.ylabel("Principle component 2")
plt.title("PCA components of digits 3,6,7")
plt.legend()
plt.show()


```

## 2. Implementing DBSCAN clustering for classes of '3', '6', and '7' of fashion-MNIST dataset:


(16) Firstly, use the DBSCAN clustering algorithm to cluster the test images directly by setting the eps parameter to 0.3 and the min_sample to 5. And then apply DBSCAN to cluster the first 2 PCA components of the test images by setting the eps parameter to 0.3 and the min_sample to 5.
Visualize the clustering results in two subplots for each k value and also print out the numbers of estimated clusters. [2 marks]






```python
##### Answer for (16)

from sklearn.cluster import DBSCAN

db_model = DBSCAN(eps = 0.3, min_samples = 5)
db_model.fit(random_test_images)
db_labels = db_model.labels_

unique_db_labels = np.unique(db_labels)

pca_model = PCA(n_components = 2)
images_pca = pca_model.fit_transform(random_test_images)
db_pca_model = DBSCAN(eps = 0.3, min_samples =5)
db_pca_model.fit(images_pca)
db_pca_labels = db_pca_model.labels_

unique_dbpca_labels = np.unique(db_pca_labels)


db_n_clusters_ = len(set(db_labels)) - (1 if -1 in db_labels else 0)
print("Clusters without PCA:", db_n_clusters_)
plt.figure(figsize=(8,6))
for label in unique_db_labels:
    mask = db_labels == label

    if label == -1:
        color = [0,0,0]
        label_name = "noise"

    else:
        color = np.random.rand(3,)
        label_name = f"Cluster {label} "

    plt.scatter(images_pca[mask, 0],images_pca[mask,1], color = color, label = label_name, alpha = 0.6)

plt.title("DBSCAN without PCA")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.legend()
plt.show()

dbpca_n_clusters = len(set(db_pca_labels)) - (1 if -1 in db_pca_labels else 0)
print("Clusters with PCA",  dbpca_n_clusters)
plt.figure(figsize=(8,6))
for label in unique_dbpca_labels:
    mask = db_pca_labels == label

    if label == -1:
        color = [0,0,0]
        label_name = "noise"
    else:
        color = np.random.rand(3,)
        label_name = f"Cluster {label} "

    plt.scatter(images_pca[mask,0], images_pca[mask,1], color = color, label = label_name, alpha = 0.6)

plt.title("DBSCAN with PCA")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.legend()
plt.show()






```

(20). Usually, DBSCAN takes longer than K-means to run, and the time it takes to run is affected by the eps parameter. Explain why that is the case. [4 marks]

Answer to (20)

During a DBSCAN, the algorithm checks every point in order to determine what type of point it is; therefore, it has to go to every point in order to determine so; however, for K-means the algorithm only has to check the distances between the point and the centroid and then update the centroid which doesnt require the algorithm to go to every point only calculate the distance making it faster. The higher the eps the higher the chance that there are going to be more points in the neighbourhood of our point, meaning that there are more points to compute, and therefore it will take longer to do so, lower eps means a lower neighbourhood radius and therefore better chances of fewer points.

(22) Load "train-labels-idx1-ubyte.gz" and "train-images-idx3-ubyte.gz". In the variable X_tmp (see the code below), every row represents a greyscale image, and every one of the 784 columns represent a pixel value. 784 = 28 x 28. So each image is size 28*28.

First, load the two files "train-labels-idx1-ubyte.gz" and "train-images-idx3-ubyte.gz". Then, perform train-test-split to prepare both the training data in terms of (X_train, y_train) and test data in terms of (X_test, y_test). The code for this question is provided below to faciliate the implementation for the following questions. [0 mark]

```python
# answer for (22). You may need to change the path for "train-labels-idx1-ubyte.gz" and "train-images-idx3-ubyte.gz".
import pandas as pd
from sklearn.model_selection import train_test_split
from matplotlib import pyplot as plt
import numpy as np


def load_mnist(path, kind='train'):
    import os
    import gzip
    import numpy as np

    """Load fashion-MNIST data from `path`"""
    labels_path = os.path.join(path,
                               '%s-labels-idx1-ubyte.gz'
                               % kind)
    images_path = os.path.join(path,
                               '%s-images-idx3-ubyte.gz'
                               % kind)

    with gzip.open(labels_path, 'rb') as lbpath:
        labels = np.frombuffer(lbpath.read(), dtype=np.uint8,
                               offset=8)

    with gzip.open(images_path, 'rb') as imgpath:
        images = np.frombuffer(imgpath.read(), dtype=np.uint8,
                               offset=16).reshape(len(labels), 784)

    return images, labels


X_tmp, y_tmp = load_mnist('t10k files', kind='train')

# normalization
X_tmp = X_tmp/255.0


plt.figure(figsize=(15,3))
for i in range(5):
  image = X_tmp[i,:].reshape(28,28)
  plt.subplot(1, 5, i+1)
  plt.imshow(image, cmap='gray')
  plt.axis('off')
plt.show()

X_train, X_test, y_train, y_test = train_test_split(X_tmp, y_tmp, test_size=0.2, random_state=1)
```

(23)

Implement a two-class Logistic Regression classifer for classes with labels 4 and 6 in fashion-MNIST by using LogisticRegression from `sklearn.linear_model`. Set random_state to 42, solver='sag' and max_iter=200 in LogisticRegression- (look at the documentation online if needed). Print both the training and test classfication accuracices. [1 mark]

```python
#answer to (23)
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score 

correct_values = [4,6]
true_values_test = np.isin(y_test, correct_values)
true_values_train = np.isin(y_train, correct_values)
image_list_train = X_train[true_values_train]
label_list_train = y_train[true_values_train]
image_list_test = X_test[true_values_test]
label_list_test = y_test[true_values_test]

LogReg = LogisticRegression(random_state =42, solver="sag", max_iter =200)

LogReg.fit(image_list_train, label_list_train)

y_pred_test = LogReg.predict(image_list_test)
y_pred_train = LogReg.predict(image_list_train)


print("Train Classification score:", accuracy_score(label_list_train, y_pred_train))
print("Test Classification score:",accuracy_score(label_list_test, y_pred_test))
```
